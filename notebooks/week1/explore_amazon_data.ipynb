```python
import json
import gzip
import pandas as pd
from pathlib import Path
```

```python
json_file = next(Path("../../data").glob("*.jsonl"))

with json_file.open("r", encoding="utf-8") as file:
    first_record = json.loads(file.readline())

print(first_record)
```

```python
df = pd.read_json(json_file, lines=True, nrows=20)
pd.reset_option("display.max_colwidth")
df
```

```python
import matplotlib.pyplot as plt

df = pd.read_json(json_file, lines=True)  # Load all rows, not nrows=20

df["average_rating"] = pd.to_numeric(df["average_rating"], errors="coerce")

category_df = df.explode("categories")

groups = [
    (df, "main_category"),
    (category_df, "categories"),
    (df, "store"),
]

fig, axes = plt.subplots(3, 1, figsize=(10, 15))

for ax, (data, column) in zip(axes, groups):
    averages = (
        data.groupby(column)["average_rating"]
        .agg(["mean", "count"])
        .nlargest(15, "count")
        .sort_values("mean")
    )

    averages["mean"].plot(kind="barh", ax=ax)
    ax.set_title(f"Average rating by {column} (15 largest groups)")
    ax.set_xlabel("Average rating")
    ax.set_ylabel(column)

plt.tight_layout()
plt.show()
```

```python
price = pd.to_numeric(
    df["price"].astype(str).str.replace(r"[$,]", "", regex=True),
    errors="coerce",
)
rating = pd.to_numeric(df["average_rating"], errors="coerce")

plot_data = pd.DataFrame({"price": price, "rating": rating}).dropna()

print(f"Pearson correlation:  {plot_data['price'].corr(plot_data['rating'], method='pearson'):.3f}")
print(f"Spearman correlation: {plot_data['price'].corr(plot_data['rating'], method='spearman'):.3f}")

plot_data.plot.scatter(x="price", y="rating", alpha=0.3, figsize=(8, 5))
plt.title("Product rating vs. price")
plt.show()
```

```python
price_cutoff = plot_data["price"].quantile(0.99)
filtered_data = plot_data[plot_data["price"] <= price_cutoff]

print(f"Price cutoff: {price_cutoff:.2f}")
print(f"Removed {len(plot_data) - len(filtered_data)} products")

print(f"Pearson correlation:  {filtered_data['price'].corr(filtered_data['rating'], method='pearson'):.3f}")
print(f"Spearman correlation: {filtered_data['price'].corr(filtered_data['rating'], method='spearman'):.3f}")

filtered_data.plot.scatter(x="price", y="rating", alpha=0.3, figsize=(8, 5))
plt.title("Product rating vs. price (excluding top 1% of prices)")
plt.show()
```

```python
pearson = filtered_data["price"].corr(filtered_data["rating"], method="pearson")
spearman = filtered_data["price"].corr(filtered_data["rating"], method="spearman")

print(f"Pearson correlation:  {pearson:.3f}")
print(f"Spearman correlation: {spearman:.3f}")
```
